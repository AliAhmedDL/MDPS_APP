```python
# Imporing The Dependencies

import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import f1_score, make_scorer
from xgboost import XGBClassifier
from sklearn.linear_model import LogisticRegression
import pickle
```

```python
# Loading the dataset to pandas DataFrame

diabetes_data = pd.read_csv('D:\Projects\MultipleDiseasePredictionSystemApp\diabetes.csv')
```

```python
# Printing the first 5 rows of the data set

diabetes_data.head()
```

```python
# Printing information on the data set

diabetes_data.info()
```

```python
# Checking for missing values

diabetes_data.isnull().sum()
```

```python
# Separating the data and labels

X = diabetes_data.drop(columns = ['Outcome'], axis = 1)
Y = diabetes_data['Outcome']
```

```python
# Data Standrization

scaler = StandardScaler()
scaler.fit(X)
standarized_data = scaler.transform(X)
```

```python
# Initializing the LogisticRegreesion model

Logistic_model = LogisticRegression()

# Define the hyperparameter grid

hyperparameters = {
    'penalty': ['l1', 'l2'],
    'C': [0.001, 0.01, 0.1, 1, 10, 100],
    'solver': ['newton-cg', 'lbfgs', 'liblinear', 'sag', 'saga'],
    'max_iter': [100, 200, 300, 400, 500]
}

# Define the F1 scorer

f1_scorer = make_scorer(f1_score)

# Using grid search to get the best hyperparameters for our model 

grid_search = GridSearchCV(Logistic_model ,hyperparameters, cv=5, scoring = f1_scorer)
grid_search.fit(standarized_data, Y)

# Intiallizing the model with the best hyper parameters
Logistic_model = LogisticRegression(penalty = grid_search.best_params_['penalty'], C = grid_search.best_params_['C'], solver = grid_search.best_params_['solver'],  max_iter = grid_search.best_params_['max_iter'])
```

```python
# Initializing the SVM model

SVM_model = svm.SVC()

# Define the hyperparameter grid

hyperparameters = {
    'C': [0.1, 1, 10, 100],
    'gamma': [1, 0.1, 0.01, 0.001],
    'kernel': ['rbf', 'linear', 'poly','sigmoid'],
}

# Define the F1 scorer

f1_scorer = make_scorer(f1_score)

# Using grid search to get the best hyperparameters for our model 

grid_search = GridSearchCV(SVM_model ,hyperparameters, cv=5, scoring = f1_scorer)
grid_search.fit(standarized_data, Y)

# Intiallizing the model with the best hyper parameters

SVM_model = svm.SVC(C = grid_search.best_params_['C'], gamma = grid_search.best_params_['gamma'], kernel = grid_search.best_params_['kernel'])
```

```python
# Intiallizing the second model XGBClassifier

XGB_model = XGBClassifier()

# Define the hyperparameter grid

hyperparameters = {
    'learning_rate': [0.1, 0.01, 0.001],
    'max_depth': [3, 5, 7],
    'subsample': [0.8, 0.9, 1.0]
}

# Define the F1 scorer

f1_scorer = make_scorer(f1_score)

# Using grid search to get the best hyperparameters for XGBClassifier model 

grid_search = GridSearchCV(XGB_model ,hyperparameters, cv=5, scoring = f1_scorer)
grid_search.fit(standarized_data, Y)

# Intiallizing the model with the best hyper parameters
XGB_model = XGBClassifier(learning_rate = grid_search.best_params_['learning_rate'], max_depth = grid_search.best_params_['max_depth'], subsample = grid_search.best_params_['subsample'])
```

```python
# Train Test Split

X_train, X_test, Y_train, Y_test = train_test_split(standarized_data, Y, test_size = 0.2, stratify = Y, random_state = 2)
```

```python
# Fitting the models

XGB_model.fit(X_train, Y_train)
SVM_model.fit(X_train, Y_train)
Logistic_model.fit(X_train, Y_train)
```

```python
# Making Function to select the best model

def Model_selection(Model_A, Model_B, Model_C):

    # Variable to save the best model and Score
    Best_score = 0
    Best_model = 0
    # Comparing between the models using F1 score metric

    # Calculating The F1 score for SVM model
    predictions_model_A = Model_A.predict(X_test)
    F1_score_model_A = f1_score(Y_test, predictions_model_A)
    if (F1_score_model_A > Best_score):
        Best_score = F1_score_model_A
        Best_model = Model_A

    # Calculating The F1 score for Logistic Regression model
    predictions_model_B = Model_B.predict(X_test)
    F1_score_model_B = f1_score(Y_test, predictions_model_B)
    if (F1_score_model_B > Best_score):
        Best_score = F1_score_model_B
        Best_model = Model_B

    # Calculating The F1 score for XGBRegressor model
    predictions_model_C = Model_C.predict(X_test)
    F1_score_model_C = f1_score(Y_test, predictions_model_C)
    if (F1_score_model_C > Best_score):
        Best_score = F1_score_model_C
        Best_model = Model_C

    # Return the best Model

    return (Best_model, F1_score_model_A, F1_score_model_B, F1_score_model_C, Best_score)
```

```python
(Final_model, score_A, score_B, score_C, Best_score) = Model_selection(SVM_model, Logistic_model, XGB_model)
print('The F1 score of model A is: {}\n the F1 score of model B is: {}\n the F1 score of model C is: {}\n The best score is: {}'.format(score_A, score_B,score_C, Best_score))
```

```python
# Making a predictive system with the best model

input_data = (5,166,72,19,175,25.8,0.587,51)

# changing the input_data to numpy array
input_data_as_numpy_array = np.asarray(input_data)

# reshape the array as we are predicting for one instance
input_data_reshaped = input_data_as_numpy_array.reshape(1,-1)

# standardize the input data
std_data = scaler.transform(input_data_reshaped)
print(std_data)

prediction = Final_model.predict(std_data)
print(prediction)

if (prediction[0] == 0):
  print('The person is not diabetic')
else:
  print('The person is diabetic')
```
